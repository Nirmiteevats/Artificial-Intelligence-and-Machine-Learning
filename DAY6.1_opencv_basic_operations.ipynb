```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
#read and display image
image= cv2.imread('img2.jpeg')
print(image)
image.shape
```

```python
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

plt.imshow(image_rgb)
plt.title('Original Image')
plt.axis('off')
```

```python
#convert to gray scale
gray = cv2.cvtColor (image, cv2.COLOR_BGR2GRAY)

plt.imshow(gray, cmap='gray')
plt.title('Grayscale Image')
plt.axis('off')
gray.shape
```

```python
#resize image
resized = cv2.resize(image, (300, 300))
resized_rgb = cv2.cvtColor(resized, cv2.COLOR_BGR2RGB)

plt.imshow(resized_rgb)
plt.title('Resized Image')
plt.axis('off')
```

```python
#apply gaussian blur
blur = cv2.GaussianBlur(image, (5, 5), 0)
blur_rgb= cv2.cvtColor(blur, cv2.COLOR_BGR2RGB)

plt.imshow(blur_rgb)
plt.title('Blurred Image')
plt.axis('off')
```

```python
#edge detection
edges = cv2.Canny(gray, 100, 200)

plt.imshow(edges, cmap='gray')
plt.title('Edge Detection')
plt.axis('off')
```

```python
# thresholding
_, thresh = cv2.threshold(gray, 127, 255, cv2.THRESH_BINARY)

plt.imshow(thresh, cmap='gray')
plt.title('Threshold Image')
plt.axis('off')
```

```python
#save image
cv2.imwrite('output.jpg',thresh)
```

```python
import cv2
import numpy as np

# Read the image
image = cv2.imread("img2.jpeg")

# Convert to grayscale
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# Create an empty binary image
binary = np.zeros(gray.shape, dtype=np.uint8)

# Threshold value
threshold = 127

# Traverse the grayscale image
rows, cols = gray.shape

for i in range(rows):
    for j in range(cols):
        if gray[i, j] > threshold:
            binary[i, j] = 255   # White
        else:
            binary[i, j] = 0     # Black

# Save the binary image
cv2.imwrite("binary_img.jpg", binary)

# Display images
cv2.imshow("Grayscale Image", gray)
cv2.imshow("Binary Image", binary)

cv2.waitKey(0)
cv2.destroyAllWindows()
```
