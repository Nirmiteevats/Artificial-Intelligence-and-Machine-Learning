```python
import numpy as np
# Coefficient matrix
A = np.array([
    [10, 5, -6],
    [2, 2, -6],
    [3, -3, 1]
])
# Constant matrix
B = np.array([10, -5, -12])
# Solve the system
solution = np.linalg.solve(A, B)

print("Solution:")
print("x =", solution[0])
print("y =", solution[1])
print("z =", solution[2])
```

```python
import numpy as np
# Coefficient matrix
A = np.array([
    [10, 5, -6],
    [2, 2, -6],
    [3, -3, 1]
])
print(A)
# Constant matrix
B = np.array([10, -5, -12])
print(B)
# Solve the system
invA=np.linalg.inv(A)
X=np.matmul(invA,B)
print(X)
```

```python
print(np.linalg.solve(A,B))
```

```python
#Solve using cramers rule
A1=np.vstack((B,A[:,1],A[:,2]))
print(A1)
A2=np.vstack((A[:,0],B,A[:,2]))
A3=np.vstack((A[:,0],A[:,2],B))
x=np.linalg.det(A1)/np.linalg.det(A)
y=np.linalg.det(A2)/np.linalg.det(A)
z=np.linalg.det(A3)/np.linalg.det(A)
print(x,y,z)
```
