```python
#1
import numpy as np

arr = np.arange(1, 21)

print("Array:", arr)
print("Shape:", arr.shape)
print("Datatype:", arr.dtype)
```

```python
#2
zeros = np.zeros((3,3))
ones = np.ones((4,4))
identity = np.eye(5)

print("Zeros:\n", zeros)
print("\nOnes:\n", ones)
print("\nIdentity:\n", identity)
```

```python
#3
arr1 = np.arange(10,101,10)
arr2 = np.linspace(0,1,5)

print(arr1)
print(arr2)
```

```python
#4
arr = np.random.randint(1,51,(3,3))

print(arr)
```

```python
#5
arr = np.random.randint(1,100,10)

print("Original:", arr)

arr = arr.astype(float)

print("Float:", arr)
```

```python
#6
a = np.array([10,20,30])
b = np.array([2,4,5])

print("Addition:", a+b)
print("Subtraction:", a-b)
print("Multiplication:", a*b)
print("Division:", a/b)
```

```python
#7
arr = np.array([1,4,9,16])

print("Square:", np.square(arr))
print("Square Root:", np.sqrt(arr))
print("Exponential:", np.exp(arr))
print("Log:", np.log(arr))
```

```python
#8
arr = np.array([5,10,15,20,25])

print("Sum:", np.sum(arr))
print("Mean:", np.mean(arr))
print("Max:", np.max(arr))
print("Min:", np.min(arr))
```

```python
#9
A = np.array([[1,2],[3,4]])
B = np.array([[5,6],[7,8]])

print("Matrix Multiplication:\n", A @ B)

print("\nElement-wise Multiplication:\n", A * B)
```

```python
#10
arr = np.array([2,4,6,8])

print("Add 5:", arr+5)
print("Multiply by 3:", arr*3)
```

```python
#11
arr = np.arange(1,13)

print(arr.reshape(3,4))

print()

print(arr.reshape(4,3))
```
