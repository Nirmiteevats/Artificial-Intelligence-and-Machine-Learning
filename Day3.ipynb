```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
mystudentdata=pd.read_excel("Uni_Data.xlsx")
df=pd.DataFrame(mystudentdata)
df.head()
```

```python
x=df[["Input"]]
y=df[["Output"]]
```

```python
plt.plot(x,y,'ro')
plt.xlabel("Input_Hours")
plt.ylabel("Output_Hours")
```

```python
from sklearn.model_selection import train_test_split
xtrain,xtest,ytrain,ytest=train_test_split(x,y,test_size=0.3, random_state=42)
print(xtrain,ytrain)
```

```python
#Linear regression
from sklearn.linear_model import LinearRegression
lr=LinearRegression()
lr.fit(xtrain,ytrain) #model build
```

```python
#model testing
yp=lr.predict(xtest)
plt.plot(xtest,yp,color='green')
plt.scatter(xtest,ytest)
```

```python
from sklearn import metrics
Mae= metrics.mean_absolute_error(ytest,yp)
MSE= metrics.mean_squared_error(ytest,yp)
RMSE=np.sqrt(MSE)
print("MAE",Mae,"MSE",MSE, "RMSE",RMSE)
```

```python
hours1=np.array([8,4.5,6.2,3.6])
hourdf=pd.DataFrame(hours1)
yp=lr.predict(hourdf)
comb=np.hstack((hourdf,yp))
print("Number of Hours\t","Marks Obtained")
print(comb)
```
