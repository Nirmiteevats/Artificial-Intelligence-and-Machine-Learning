```python
# removing duplicates and null values
# handling null values 
# 1. drop values using-
#        data.dropna(inplace=True)
# 2. replace values-
#        numerical columns>>>median
#        categorical columns>>>mode
```

```python
print(data)
```

```python
data["Age"].fillna(data["Age"].median(),inplace=True)
data["Income"].fillna(data["Income"].median(),inplace=True)
```

```python
print(data)
```

```python
data["Age"].fillna(data["Age"].mean(),inplace=True)
data["Income"].fillna(data["Income"].mean(),inplace=True)
```

```python
print(data)
```

```python
data.head(7)
```

```python
print(data["Gender"].mode())
```

```python
print(data["City"].mode())
```

```python
data["Gender"].fillna(data["Gender"].mode()[0],inplace=True)
data["City"].fillna(data["City"].mode()[0],inplace=True)
data["Date_of_Joining"].fillna("01-01-2020",inplace=True)
print(data)
```

```python
data.drop_duplicates(inplace=True)
data.head(7)
```

```python
data["Date_of_Joining"]=pd.to_datetime(data["Date_of_Joining"])
data["Gender"]=data["Gender"].astype("category")
data.info()
```

```python
Q1=data["Income"].quantile(0.75)
Q3=data["Income"].quantile(0.75)
IQR=Q3-Q1
print(Q1,Q3,IQR)
```

```python
data=(data["Income"]>Q1-1.5*IQR)&(data["Income"]<Q3+1.5*IQR)
print(data)
```

```python
data=pd.get_dummies(data,columns=["City"],drop_first=True,dtype=int)
print(data)
```
